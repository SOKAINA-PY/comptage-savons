```python
import cv2

cap = cv2.VideoCapture("savons.mp4")
if cap.isOpened():
    print("Video ouverte avec succès ✅")
else:
    print("Erreur : vidéo introuvable ❌")


ret, frame = cap.read()

if ret:
    print("Frame récupérée avec succès ✅")
    print("Dimensions :", frame.shape)
else:
    print("Impossible de lire la vidéo ❌")

cap.release()
```

```python
import cv2
import matplotlib.pyplot as plt

cap = cv2.VideoCapture("savons.mp4")

ret, frame = cap.read()

if ret:
    
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    plt.imshow(frame_rgb)
    plt.axis("off")
    plt.show()

cap.release()
```

```python
import cv2

cap = cv2.VideoCapture("savons.mp4")

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = total_frames / fps

print("FPS :", fps)
print("Nombre total de frames :", total_frames)
print("Durée de la vidéo :", round(duration, 2), "secondes")

cap.release()
```
